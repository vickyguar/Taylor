```python
from Taylor import *
import numpy as np
import math
```

```python
FIG = "figures/"
```

```python
# función exponencial (acá se ve el periodo complejo)
def f(z):
    return np.exp(z)

def S(z, N):
    s = np.zeros_like(z, dtype=complex)
    for n in range(N+1):
        s += z**n / math.factorial(n) # la representación en taylor es sum z^n/n!
    return s

make_animation(f, S, (-2,2), (-4*np.pi,4*np.pi), f"../{FIG}exp.gif")
```

```python
# función coseno
def f(z):
    return np.cos(z)

def S(z, N):
    s = np.zeros_like(z, dtype=complex)
    for n in range(N+1):
        s += ((-1)**n) * z**(2*n) / math.factorial(2*n)
    return s

make_animation(f, S, (-4*np.pi,4*np.pi), (-2,2), f"../{FIG}cos.gif")
```

```python
# ahora una que se resuelve con la serie geometrica. 
# Notar que no es analítica en todo el plano --> tenemos una ROC del tipo |z - z0| < R
def f(z):
    return 1/(1 - z)

def S(z, N):
    s = np.zeros_like(z, dtype=complex)
    for n in range(N+1):
        s += z**n
    return s

make_animation(f, S, (-3,3), (-3,3), f"../{FIG}geom.gif")
```

```python
# misma función, pero le cambiamos el centro
def f(z):
    return 1/(1 - z)

def S(z, N):
    s = np.zeros_like(z, dtype=complex)
    for n in range(N+1):
        s += (-(z-2))**n # centro del desarrollo en z = 2
    return -s

make_animation(f, S, (-3,3), (-3,3), f"../{FIG}geom_center2.gif")
```

```python
def f(z):
    return 1/(1 + z**2)

def S(z, N):
    s = np.zeros_like(z, dtype=complex)
    for n in range(N+1):
        s += ((-1)**n) * z**(2*n)
    return s

make_animation(f, S, (-2,2), (-2,2), f"../{FIG}rational.gif")
```

```python
# hacemos un desarrollo en series de laurent ahora --> vemos que la ROC forma anillos r < |z - z0| < R
def f(z):
    return 1/((z-1)*(z-3))

def S(z, N):
    s = np.zeros_like(z, dtype=complex)

    # parte principal (|z| > 1)
    for n in range(N+1):
        s += -(1/2) * z**(-(n+1))

    # parte regular (|z| < 3)
    for n in range(N+1):
        s += -(1/6) * (z/3)**n

    return s

make_animation(f, S, (-4,4), (-4,4), f"../{FIG}laurent.gif")
```
